```python
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
df['order_approved_at'] = pd.to_datetime(df['order_approved_at'])
df['order_delivered_carrier_date'] = pd.to_datetime(df['order_delivered_carrier_date'])
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])
```

```python
df['cumple'] = (df['order_delivered_carrier_date'] <= df['order_estimated_delivery_date']).astype(int)
df.head()
```

```python
de_entregas_tarde = (sum(df['cumple']) / df['cumple'].count()) * 100
de_entregas_tarde
```

```python
data_a_revisar = df[['order_id', 'customer_id', 'cumple']]
data_a_revisar
```

```python
cliente_mas_bajo = data_a_revisar.groupby('customer_id')['cumple'].apply(lambda x: (x==0).sum()).reset_index()
cliente_mas_bajo= cliente_mas_bajo.sort_values(by= 'cumple', ascending= False)
cliente_mas_bajo
```

```python
status_a_revisar = df[['order_id', 'order_status', 'cumple']]

status = status_a_revisar.groupby('order_status')['cumple'].sum().reset_index()
status = status.sort_values(by='cumple', ascending=False)
status
```

```python
status_bajo = status_a_revisar.groupby('order_status')['cumple'].apply(lambda x: (x==0).sum()).reset_index()
status_bajo = status_bajo.sort_values(by='cumple', ascending= False)
status_bajo
```

```python
status_completo = status.merge(status_bajo, on='order_status')
status_completo = status_completo.rename(columns={'cumple_y': 'Falla'})
total = status_completo['cumple_x'].sum() + status_completo['Falla'].sum()
status_completo['%de falla'] = round(status_completo['Falla'] / total * 100,2)
status_completo.sort_values(by='Falla', ascending=False)
```
